```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Convert BGR (OpenCV default) to RGB for matplotlib
test_img = cv2.imread("./20230711_234332.jpg")
test_img_rgb = cv2.cvtColor(test_img, cv2.COLOR_BGR2RGB)

plt.imshow(test_img_rgb)
plt.axis('off')  # Hide axes
plt.show()
```

```python
test_img_bw = cv2.imread("./20230711_234332.jpg", cv2.IMREAD_GRAYSCALE)
plt.imshow(test_img_bw, cmap='gray')
plt.axis('off')
plt.show()
```

```python
test_img_bw_neg = 255 - test_img_bw
plt.imshow(test_img_bw_neg, cmap='gray')
plt.axis('off')
plt.show();
```

```python
test_img_neg = 255 - test_img
plt.imshow(test_img_neg)
plt.axis('off')
plt.show();
```

```python
b, g, r = cv2.split(test_img)
test_img_blue = cv2.merge([np.zeros_like(b), np.zeros_like(b), b])
test_img_green = cv2.merge([np.zeros_like(b), g, np.zeros_like(b)])
test_img_red = cv2.merge([r, np.zeros_like(b),  np.zeros_like(b)])
fig, axes = plt.subplots(1, 3, figsize=(30, 10))
axes[0].imshow(test_img_blue)
axes[0].axis('off')
axes[1].imshow(test_img_green)
axes[1].axis('off')
axes[2].imshow(test_img_red)
axes[2].axis('off')
plt.show()
```

```python
hsv_img = cv2.cvtColor(test_img, cv2.COLOR_BGR2HSV)
b, g, r = cv2.split(hsv_img)
fig, axes = plt.subplots(1, 3, figsize=(30, 10))
axes[0].imshow(b, cmap='grey')
axes[0].axis('off')
axes[1].imshow(g, cmap='grey')
axes[1].axis('off')
axes[2].imshow(r, cmap='grey')
axes[2].axis('off')
plt.show()
```

```python
gamma = 2
gamma_adjusted_img = cv2.cvtColor(test_img, cv2.COLOR_BGR2RGB)
gamma_adjusted_img = np.array(255*(gamma_adjusted_img/255)**gamma, dtype='uint8')
plt.imshow(gamma_adjusted_img)
```
